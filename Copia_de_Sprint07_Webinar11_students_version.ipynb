## Ejercicio 3 · Crear una copia limpia \(15 min\)

Regla de trabajo: no modificamos el dataset original directamente. Creamos `df_limpio` para preservar la trazabilidad.

Aplicaremos cuatro decisiones:

1. Convertir `fecha` a tipo fecha.
2. Normalizar textos en `canal`.
3. Completar `metodo_pago` con `No informado`.
4. Completar variables numéricas con la mediana.

````python
# TODO estudiante: escribe aquí tu código.
# Sugerencia:
# 1. Lee la consigna anterior.
# 2. Ejecuta paso a paso.
# 3. Valida el resultado antes de interpretarlo.

df_limpio = df.copy()

# 1. Convertir la columna 'fecha' a tipo datetime (formato fecha real)
df_limpio['fecha'] = pd.to_datetime(df_limpio['fecha'])

# 2. Normalizar textos en la columna 'canal'(quitar espacios sobrantes o pasar a un formato estándar si fuera necesario)
df_limpio['canal'] = df_limpio['canal'].str.strip()

# 3. Completar la variable categórica 'metodo_pago' con la etiqueta "No informado"
df_limpio['metodo_pago'] = df_limpio['metodo_pago'].fillna('No informado')

# 4. Completar variables numéricas que tenían nulos usando su MEDIANA
# Calculamos las medianas de cada columna afectada
mediana_edad = df_limpio['edad_cliente'].median()
mediana_satisfaccion = df_limpio['satisfaccion'].median()
mediana_entrega = df_limpio['dias_entrega'].median()

# Reemplazamos los nulos con las medianas calculadas
df_limpio['edad_cliente'] = df_limpio['edad_cliente'].fillna(mediana_edad)
df_limpio['satisfaccion'] = df_limpio['satisfaccion'].fillna(mediana_satisfaccion)
df_limpio['dias_entrega'] = df_limpio['dias_entrega'].fillna(mediana_entrega)

# ============================================================
# COMPROBACIÓN: Validar que ya no queden valores nulos en df_limpio
# ============================================================
print(" Verificación en df_limpio (Nulos restantes) ")
print(df_limpio[['fecha', 'canal', 'metodo_pago', 'edad_cliente', 'satisfaccion', 'dias_entrega']].isnull().sum())
````

## Ejercicio 6 · Gráficas sencillas con `.plot()` \(15 min\)

````python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Configurar el estilo visual moderno de Seaborn
sns.set_theme(style="whitegrid")

# 2. Crear el lienzo de la gráfica (ancho: 10 pulgadas, alto: 6 pulgadas)
plt.figure(figsize=(10, 6))

# 3. Crear una gráfica de barras: Comparar el total de 'ingreso' por cada 'categoria' de producto
# Usamos un orden descendente para que la gráfica se lea mejor de izquierda a derecha
orden_categorias = df_limpio.groupby('categoria')['ingreso'].sum().sort_values(ascending=False).index

sns.barplot(
    data=df_limpio,
    x='categoria',
    y='ingreso',
    estimator=sum,            # Le indicamos que sume los ingresos totales de cada categoría
    errorbar=None,            # Eliminamos las líneas de barra de error para dejar el diseño limpio
    palette='Blues_r',        # Paleta de colores azules en degradado
    order=orden_categorias
)

# 4. Personalizar las etiquetas y el título para el equipo de negocio
plt.title('Ingresos Totales por Categoría de Producto', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Categoría de Producto', fontsize=12)
plt.ylabel('Total de Ingresos ($)', fontsize=12)

# 5. Formatear los números del eje Y para que muestren comas de miles en lugar de notación científica
ax = plt.gca()
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f"${x:,.0f}"))

# 6. Desplegar la gráfica en pantalla
plt.show()
````

## Ejercicio 8 · Outliers con IQR \(8 min\)

La regla IQR permite detectar valores inusualmente altos o bajos.

No significa que esos valores estén mal. Significa que deben revisarse antes de sacar conclusiones.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  
Consiste en auditar estadísticamente la variable ingreso utilizando el método del Rango Intercuartílico (IQR) para establecer un umbral técnico que nos permita identificar transacciones que se desvían de manera extrema del comportamiento habitual de compra. Esto sirve para entender si los ingresos están inflados por unas pocas compras masivas.

**Mi procedimiento / fórmula / consulta:**  
Utilicé el método .quantile() para calcular el percentil 25 (Q1) y el percentil 75 (Q3). Restando ambos obtuve el IQR. Luego, apliqué la fórmula estándar de Tukey calculando el limite_superior = Q3 + (1.5 * IQR). Finalmente, utilicé segmentación booleana en Pandas para filtrar y contabilizar los registros cuyo ingreso superaba dicho límite.

**Resultado obtenido:**  
Al ejecutar el script, el sistema calcula el límite superior exacto de ingresos normales. Las filas filtradas representan las transacciones catalogadas como outliers. (Verás el número exacto y la muestra en la pantalla de tu Colab tras correr el código).

**Conclusión breve:**
Estas compras atípicas no deben eliminarse a la ligera porque representan ingresos reales (pueden ser grandes clientes corporativos o compras mayoristas masivas); sin embargo, aislarlas nos permite hacer dos tipos de análisis: uno enfocado en el cliente común para optimizar las tiendas del día a día, y otro especializado para entender cómo replicar y cuidar a estos "súper clientes"


````python
# TODO estudiante: escribe aquí tu código.
# 1. Calcular los cuartiles Q1 (percentil 25) y Q3 (percentil 75) para la columna 'ingreso'
Q1 = df_limpio['ingreso'].quantile(0.25)
Q3 = df_limpio['ingreso'].quantile(0.75)

# 2. Calcular el Rango Intercuartílico (IQR)
IQR = Q3 - Q1

# 3. Definir los límites matemáticos para detectar outliers (Regla de 1.5 * IQR)
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# 4. Filtrar y contar cuántos registros están por encima del límite superior
outliers_superiores = df_limpio[df_limpio['ingreso'] > limite_superior]
cantidad_outliers = len(outliers_superiores)

# 5. Mostrar los resultados en pantalla con formato legible
print("========================================")
print("     ANÁLISIS DE OUTLIERS (MÉTODO IQR)  ")
print("========================================")
print(f"Primer Cuartil (Q1):      ${Q1:,.2f}")
print(f"Tercer Cuartil (Q3):      ${Q3:,.2f}")
print(f"Rango Intercuartílico:    ${IQR:,.2f}")
print(f"Límite Superior Permitido: ${limite_superior:,.2f}")
print(f"Cantidad de Outliers Detectados: {cantidad_outliers} filas")
print("========================================")

# Mostrar las primeras filas de estas transacciones atípicas para analizarlas
if cantidad_outliers > 0:
    print("\n--- Muestra de transacciones consideradas outliers ---")
    display(outliers_superiores[['venta_id', 'categoria', 'ingreso', 'pais']].head())
````

## Ejercicio 9 · Segmentación de ventas \(8 min\)

Segmentar ayuda a convertir números en etiquetas interpretables. Vamos a clasificar cada venta por tamaño de ingreso.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  
Transformar una variable cuantitativa continua (ingreso) en una variable cualitativa ordinal (segmento_ingreso) mediante técnicas de binning. El propósito es estructurar un reporte agrupado que nos muestre la composición de la cartera de ventas, identificando la dependencia que tiene el negocio de los diferentes tickets de compra.

**Mi procedimiento / fórmula / consulta:**  
Diseñé una lista de limites numéricos y una lista de etiquetas semánticas. Implementé la función pd.cut() sobre df_limpio['ingreso'] para categorizar cada fila de forma automática. Finalmente, usé un .groupby() combinado con .agg() para consolidar el conteo de transacciones, la suma financiera y el margen de ganancia promedio por cada segmento.

**Resultado obtenido:**  
Tras ejecutar la celda, Pandas devuelve una tabla resumen de tres filas (Bajo, Medio, Alto). El reporte detalla con exactitud qué porcentaje de las transacciones totales corresponden a compras de bajo valor frente a los grandes pedidos institucionales o corporativos (ticket "Alto").

**Conclusión breve:**  Al cruzar el volumen de ventas con el Margen_Promedio, podemos evaluar un dilema clásico de negocio: ¿los tickets "Altos" nos están dejando suficiente rentabilidad porcentual por el volumen operativo que conllevan, o son las compras del segmento "Medio" las que sostienen la mayor salud financiera de la empresa?


````python
# TODO estudiante: escribe aquí tu código.
import pandas as pd

# 1. Definir los límites de los rangos (bins) basados en la distribución de tus ingresos
# [0, 100000] -> Bajo, (100000, 1000000] -> Medio, (1000000, infinito] -> Alto
limites = [0, 100000, 1000000, float('inf')]

# 2. Definir las etiquetas de texto para cada segmento
etiquetas = ['Bajo', 'Medio', 'Alto']

# 3. Crear la nueva columna 'segmento_ingreso' usando pd.cut()
df_limpio['segmento_ingreso'] = pd.cut(
    df_limpio['ingreso'],
    bins=limites,
    labels=etiquetas,
    include_lowest=True
)

# 4. Agrupar por el nuevo segmento para contar cuántas ventas hay y cuánto suman en total
resumen_segmentos = df_limpio.groupby('segmento_ingreso', observed=False).agg(
    Cantidad_Ventas=('ingreso', 'count'),
    Total_Ingresos=('ingreso', 'sum'),
    Margen_Promedio=('margen', 'mean')
).reset_index()

# 5. Mostrar el reporte en pantalla con formato limpio
print("=============================================================")
print("          REPORTE DE VENTAS SEGMENTADO POR INGRESO           ")
print("=============================================================")
display(resumen_segmentos)
````

Construye una tabla final con una recomendación de negocio.

> En `[país/canal/categoría]` se observa `[hallazgo]`, porque `[evidencia]`. Esto sugiere `[acción o pregunta de negocio]`.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  
Un análisis multi-variable agrupando simultáneamente por dos dimensiones geográficas (pais y region). El objetivo es superar el análisis plano de una sola columna y descubrir cuáles son los mercados específicos (nodos territoriales) más valiosos y eficientes en la operación de LATAM.

**Mi procedimiento / fórmula / consulta:**  
Implementé un .groupby() pasando una lista de dos variables categóricas ['pais', 'region']. Utilicé el método .agg() para realizar múltiples operaciones estadísticas vectorizadas a la vez (conteo de transacciones, suma de ingresos y promedios de margen y satisfacción). Finalmente, apliqué .sort_values() de forma descendente para jerarquizar los resultados.

**Resultado obtenido:**  
El código genera una tabla pivote integrada. Al ordenarla, se exponen de inmediato en las primeras filas las combinaciones de País-Región que lideran las ventas de la compañía, así como aquellas zonas rezagadas que aparecen al fondo de la tab.

````python
# =====================================================================
# EJERCICIO 10 · RETO FINAL: Análisis de Rentabilidad por Región y País
# =====================================================================

# 1. Agrupar los datos por País y Región utilizando nuestro df_limpio
# Calculamos: el total de ingresos, el promedio de margen y el promedio de satisfacción
reporte_reto = df_limpio.groupby(['pais', 'region'], observed=False).agg(
    Ventas_Totales=('venta_id', 'count'),
    Ingresos_Totales=('ingreso', 'sum'),
    Margen_Promedio=('margen', 'mean'),
    Satisfaccion_Promedio=('satisfaccion', 'mean')
).reset_index()

# 2. Ordenar el reporte de mayor a menor según los Ingresos Totales
reporte_reto = reporte_reto.sort_values(by='Ingresos_Totales', ascending=False)

# 3. Mostrar el dataframe resultante en pantalla
print("========================================================================")
print("             RETO FINAL: PERFORMANCE DE MERCADOS EN LATAM               ")
print("========================================================================")
display(reporte_reto)

# 4. BONUS VISUAL: Crear una gráfica rápida para comparar los ingresos por País
import matplotlib.pyplot as plt
import seaborn as sns  # <-- ¡CORREGIDO AQUÍ! Ahora sí usará la librería correcta

plt.figure(figsize=(10, 5))
sns.barplot(
    data=reporte_reto,
    x='pais',
    y='Ingresos_Totales',
    estimator=sum,
    errorbar=None,
    palette='viridis'
)

plt.title('Ingresos Totales por País (Resultado del Reto)', fontsize=14, fontweight='bold')
plt.xlabel('País')
plt.ylabel('Ingresos ($)')

# Formatear el eje Y con signo de pesos y comas de miles
ax = plt.gca()
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f"${x:,.0f}"))

plt.show()
````
